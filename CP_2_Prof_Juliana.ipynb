1. O que é um Perceptron? (1,0 ponto)

Resposta:
O Perceptron é um modelo básico de rede neural artificial criado para realizar tarefas de classificação. Ele recebe entradas, realiza cálculos com pesos e produz uma saída baseada em uma função de ativação.
____________________________________________________________
2. Quais são os principais elementos que compõem um Perceptron? (1,0 ponto)

Resposta:
São as entradas (x), os pesos (w), o viés (bias), o somatório ponderado e a função de ativação.
____________________________________________________________
3. Qual é a função dos pesos dentro do Perceptron? (1,0 ponto)

Resposta:
Os pesos determinam a importância de cada entrada no resultado final. Durante o treinamento, eles são ajustados para reduzir os erros do modelo.
____________________________________________________________
4. Qual é a função da função de ativação? (1,0 ponto)

Resposta:
A função de ativação decide qual será a saída do Perceptron com base no valor calculado pelo somatório das entradas e dos pesos.
